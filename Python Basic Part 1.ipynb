# 31. GCD Calculator

```python
# Euclidean Algorithm
n1=60
n2=48

while n2!=0:
    n1,n2=n2,n1%n2

print(abs(n1))
```

```python
import math

n1=60
n2=48

result= math.gcd(n1,n2)

print(f"The GCD of {n1} and {n2} is : {result}")
```

```python
# Recursive method Euclidean
def gcd_calc(a,b):
    if b ==0:
        return  abs(a)
    else:
        return gcd_calc(b,a%b)

print(gcd_calc(60,48))    
```
