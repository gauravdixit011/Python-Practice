# 31. GCD Calculator

```python
# Write a Python program that computes the greatest common divisor (GCD) of two positive integers.
# Method - 1
# Euclidean Algorithm

n1=60
n2=48

while n2!=0:
    n1,n2=n2,n1%n2

print(abs(n1))
```

```python
# Method - 2
import math

n1=60
n2=48

result= math.gcd(n1,n2)

print(f"The GCD of {n1} and {n2} is : {result}")
```

```python
# Method - 3
# Recursive method Euclidean
def gcd_calc(a,b):
    if b ==0:
        return  abs(a)
    else:
        return gcd_calc(b,a%b)

print(gcd_calc(60,48))    
```

# 32. LCM Calculator
Write a Python program to find the least common multiple (LCM) of two positive integers. 

```python
# Method - 1 Loop  (Brute Force)
a = 45
b = 98

greater = a if a>b else b  # greater = max(a,b)

while True:
    if greater%a==0 and greater%b==0:
        lcm=greater
        break
    greater+=1

print(f"The LCM of {a} and {b} is: {lcm}")
```

```python
# Method - 2 
# Built-in function math.lcm()  (Python 3.9+)
import math

a = 45
b = 98

lcm = math.lcm(a, b)

print(f"The LCM of {a} and {b} is: {lcm}")
```

```python
# Method - 3  using the gcd farmula (Prefer for Interviews)
import math 

a = 45
b = 98

lcm = abs(a*b) // math.gcd(a, b)

print(f"The LCM of {a} and {b} is: {lcm}")
```

# 42. Shell Mode Detector

Write a Python program to determine if a Python shell is executing in 32bit or 64bit mode on OS.

```python
# You can use Python's platform.architecture() 
# to determine whether the current Python interpreter 
# is running in 32-bit or 64-bit mode.

import platform

mode = platform.architecture()[0]

print("Python Shell Mode:", mode)
```

```python
# Better approach

# For checking the actual interpreter architecture, 
# struct.calcsize() is a more direct method:

import struct

bits = struct.calcsize("P") * 8

print(f"Python Shell is running in {bits}-bit mode")
```

Why this works: "P" represents a pointer, 

and its size is typically 4 bytes on a 32-bit interpreter 

and 8 bytes on a 64-bit interpreter.
